TRANSFORMATION API
RDD를 새로운 RDD로 변형(read-only 특성)
RDD Lineage
 - trasformation 이 적용 되는 연산 순서를 나타내는 DAG
 - memory에서 유실 되더라도 DAG를 따라서 데이터 복구 간응
Lazy-Evaluation 떄문에 transformation 연산만 호출하면 실제 연산이
일어나지 않음

함수 종류
map() : 각 row에 함수를 적용하고 결과 RDD 반환
filter() : 조건을 만족하는 row만 남겨서 반환
union() : 두 RDD 의 합집합 RDD 반환
distinct() : RDD 내의 중복값을 제거한 RDD 반환

TRANSFORMATION API
실제 연산을 발생시키는 명령
리턴 값이 데이터 이거나 실행 결과
Action에 해당하는 API가 실행되기 전까지 실제 연산은 수행되지 않음

함수 종류
collect() : RDD의 모든 값 리턴
count() : RDD의 값 개수 연산 리턴
reduce(f) : reduce(f) 병렬로 f 함수에 해당하는 병합 연산 리턴
take(n) : 상위 n 만큼의 개수를 리턴
