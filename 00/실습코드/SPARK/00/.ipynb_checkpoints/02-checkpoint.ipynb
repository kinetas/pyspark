```python
# Spark Context를 생성해야 spark task를 실행 가능
# Initialize Environment 역할
# low-level에서 작업하기 위해 session 사용 안함

from pyspark import SparkConf, SparkContext

sc = SparkContext("local","word-count")

# read함수 보다 더 low-level textFile 파이선 모델을 스파크에서 사용할 수 있는 데이터
# 모델로 변환시켜 사용
text = sc.textFile("text_file_path")
words = text.flatMap(lambda line:line.split(" "))

# function map(String key, String document_contents):
#     # 입력: key(파일위치/라인번호), document_contents(텍스트 본문)
#     # 처리: 본문에서 단어를 하나씩 꺼내는 루프(for each)를 돌림
    
#     for each String w in document_contents.split(" "):
#         if w.isValid():
#             emit(w, 1)  # 단어 w와 숫자 1을 매퍼의 결과물로 방출(emit)
word_counts = words.map(lambda word: (word,1))
# function reduce(String w, Iterator values):
#     # 입력: 단어 w, 그리고 그 단어에 모인 1들의 묶음(values)
#     int sum = 0
    
#     # 묶여서 들어온 1들을 하나씩 꺼내서 더하는 루프(for each)
#     for each int v in values:
#         sum += v  # lambda a, b: a + b 연산이 루프를 돌며 수행됨
        
#     emit(w, sum)  # 최종 단어와 총 개수를 최종 결과물로 방출(emit)
word_counts = word_counts.reduceByKey(lambda a,b:a+b)
```
